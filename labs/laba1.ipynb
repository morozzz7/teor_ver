В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, ecли по каждой номинации установлены:
- а) различные призы;

## Номер 1.9
Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

- Рассмотрим событие $A$ - в семизначном числе, состоящем из цифр 4, 5, 6, цифра 4 повторяется 3 раза, а цифры 5 и 6 - 2 раза

```python
sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * 2)
```

## Номер 1.10
Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: 

а) 3 карточки; 

б) все 6 карточек.

Какова вероятность того, что получится слово: 

а) «ТОР»; 

б) «ТЕОРИЯ»

- Рассмотрим событие $A$ - получится слово "ТОР"

```python
round(1 / (6 * 5 *4), 5)
```

- Рассмотрим событие $B$ - получилось слово "ТЕОРИЯ"

```python
round(1 / sc.factorial(6), 4)
```

## Пример 1.11
Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна С

- Рассмотри событие $A$ - из карточек получилось слово "АНАНАС" 

```python
round(sc.factorial(3) * sc.factorial(2) / sc.factorial(6), 3)
```

## Пример 1.16
За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: 

а) никакие два лица одного пола не сядут рядом; 

б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

- Рассмотрим событие $A$ - никакие два лица одного пола не сядут рядом

```python
round(2 * (sc.factorial(5))**2 / sc.factorial(10), 5)
```

- Рассмотрим событие $B$ - супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар

```python
round(sc.factorial(5) * 2**5 / sc.factorial(10), 5)
```

## Пример 1.17
В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали:

а) в два купе;

б) в семь купе; 
 
в) в три купе

- Рассмотрим событие $A$ - пассажиры попали в 2 купе

```python
round(sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7),7)
```

- Рассмотрим событие $B$ - пассажиры попали в 7 купе

```python
round(sc.comb(9, 7) * 4**7 / sc.comb(36, 7), 5)
```
