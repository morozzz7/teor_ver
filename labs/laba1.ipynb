## Номер 1.48

В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: 

а) выпадения 11 очков; 

б) выигрыша.


- Рассмотрим событие $A$ - выпало число 11 в сумме на трех кубиках

```python
variants = list(it.product(range(1, 7), repeat=3))

right_sum = sum(1 for variant in variants if sum(variant) == 11)

right_sum / len(variants)
```

- Рассмотрим событие $B$ - выпало число больше 10

```python
var_count = sum(1 for variant in variants if sum(variant) > 10)
var_count / len(variants)
```
