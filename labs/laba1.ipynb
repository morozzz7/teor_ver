# Тема 3. Условная вероятность

Решить классически и смоделировать решение задачи. 

В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

## Классическое решение

Рассмотрим событие $A$ - шар, вытянутый из второй урны - белый

```python
w1, b1 = 15, 5
w2, b2 = 6, 1
total_1 = w1 + b1
balls_to_move = 3
total_2 = w2 + b2 + balls_to_move
final_w_prob = 0

total_comb = sc.comb(total_1, balls_to_move)
for w in range(balls_to_move + 1):
    b = balls_to_move - w
    prob_transfer = (sc.comb(w1, w) * sc.comb(b1, b)) / total_comb
    prob_w_from_2 = (w2 + w) / total_2
    final_w_prob += prob_transfer * prob_w_from_2


final_w_prob
    
```

## Моделирование решения задачи

```python

def run_simulation(trials):
    urn1 = [1] * 15 + [0] * 5  # 1 - белый, 0 - черный
    success = 0
    for _ in range(trials):
        moved = random.sample(urn1, 3)
        w2_current = 6 + sum(moved)
        if random.random() < (w2_current / 10.0):
            success += 1
    return success / trials

# Моделирование для разного количества шагов
trial_counts = [10, 50, 100, 500, 1000, 5000, 10000, 50000, 100000, 1000000]
data = []

for t in trial_counts:
    sim_p = run_simulation(t)
    deviation = abs(sim_p - final_w_prob)
    data.append({"Испытания": t, "Моделирование": sim_p, "Отклонение": deviation})

df = pd.DataFrame(data)
print(df.to_string(index=False))

plt.figure(figsize=(10, 6))
plt.plot(df["Испытания"], df["Моделирование"], marker='o', linestyle='-', color='b', label='Эксперимент')
plt.axhline(y=final_w_prob, color='r', linestyle='--', label=f'Теория ({final_w_prob})')

plt.xscale('log')
plt.xlabel('Число испытаний (N)')
plt.ylabel('Вероятность белого шара')
plt.title('Сходимость частоты эксперимента к истинной вероятности')
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины:

а) 480 предприятий; 

б) наивероятнейшее число предприятий; 

в) не менее 480; 

г) от 480 до 520.

```python
n = 1000  # число предприятий
p = 0.5  # вероятность того, что предприятие имеет нарушение фин. дисциплины
q = 1 - p
sigma = np.sqrt(n * p * q)
True if n * p * q >= 20 else False
```

$n*p*q$ ≥ 20, значит можно использовать локальную и интегральную теоремы Муавра-Лапласса

Рассмотрим событие $A$ - 480 предприятий имеют нарушения

```python
# Классический способ

prob1 = round(binom.pmf(480, n, p), 6)

# Локальная теорема Муавра-Лапласса

x = (480 - n*p) / sigma
prob2 = round(norm.pdf(x) / sigma, 6)

print(prob1)
print(prob2)
```

Рассмотрим событие $B$ - наивероятнейшее число предприятий имеют нарушения

```python
# Классический способ

lower = n * p - q
upper = n * p + q

m0_candidates = [m for m in range(n + 1) if lower <= m <= upper]

for m in m0_candidates:
    print(round(binom.pmf(m, n, p), 6))
```

Рассмотрим событие $C$ - не менее 480 предприятий имеют нарушения

```python
# Классический способ

prob1 = round(1 - binom.cdf(479, n, p), 6)

# Интегральная формула Муавра-Лапласса

x1 = (1000.5 - n*p) / sigma
x2 = (479.5 - n*p) / sigma
prob2 = round(norm.cdf(x1) - norm.cdf(x2), 6)

print(prob1)
print(prob2)
```

Рассмотрим событие $D$ - от 480 до 520 предприятий имеют нарушения

```python
# Классический способ

from scipy.stats import sigmaclip


prob1 = binom.cdf(520, n, p) - binom.cdf(479, n, p)

# Интегральная формула Муавра-Лапласса

x1 = (520.5 - n*p) / sigma
x2 = (479.5 - n*p) / sigma
prob2 = norm.cdf(x1) - norm.cdf(x2)

print(round(prob1, 6))
print(round(prob2, 6))
```
