## Номер 1.38
Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева
направо в порядке нумерации томов (от 1 до 5)?

- Рассмотри событие $A$ - книги стоят слева на право в порядке от 1 до 5

```python
round(1 / sc.factorial(5), 5)
```

## Номер 1.43

Наудачу взятый телефонный номер состоит из 5 цифр.
Какова вероятность того, что в нем все цифры:

а) различные;

б) одинаковые; 

в) нечетные

Известно, что номер телефона не начинается с цифры ноль.

- Рассмотрим событие $A$ - все цифры в номере различные

```python
9 * 9 * 8 * 7 * 6 / (9 * 10**4)
```

- Рассмотрим событие $B$ - все цифры в номере одинаковые

```python
9 / (9 * 10**4)
```

- Рассмотрим событие $C$ - все цифры в номере нечетные

```python
round(5** 5/(9 * 10**4), 5)
```
